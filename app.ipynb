```python
# to csv
df =pd.read_csv('deliveries.csv')
df
```

```python
temp_df=df.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).reset_index()
```

```python
temp_df.to_csv('batsman_runs_csv',index=False)
```

```python
df.pivot_table(index='batsman',columns='bowling_team',values='batsman_runs',aggfunc='sum')
```

### to_excel

```python
temp_df=df.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).reset_index()
```

```python
temp_df.to_excel('batsman_runs.xlsx')
```
